```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
# Load your image in grayscale
clean_image = cv2.imread('image.webp', cv2.IMREAD_GRAYSCALE)

# Display the image using Matplotlib
plt.imshow(clean_image, cmap='gray')
plt.title('1. Original Custom Image')
plt.axis('off') # Hides the axis numbers
plt.show()
```

```python
# Define noise parameters (you can change the 40 later to make it more/less noisy)
mean = 0
standard_deviation = 40 

# Generate Gaussian noise matching the image dimensions
gaussian_noise = np.random.normal(mean, standard_deviation, clean_image.shape)

# Add the noise to the original image
# We temporarily convert the image to float32 to prevent math errors during addition
noisy_image = clean_image.astype(np.float32) + gaussian_noise

# Clip values to ensure they stay within the standard image range (0-255)
noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8)

# Display the noisy image
plt.imshow(noisy_image, cmap='gray')
plt.title('2. Image with Gaussian Noise')
plt.axis('off')
plt.show()
```

```python
# Apply Gaussian Filter to the noisy image
# The (7, 7) is the kernel size. It must be odd numbers (like 3, 5, 7).
# Larger numbers = stronger blur.
filtered_image = cv2.GaussianBlur(noisy_image, (7, 7), 0)

# Display the filtered image
plt.imshow(filtered_image, cmap='gray')
plt.title('3. Filtered Image (Denoised)')
plt.axis('off')
plt.show()
```

```python
# Set up a figure with 1 row and 3 columns
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Plot Original
axes[0].imshow(clean_image, cmap='gray')
axes[0].set_title('1. Original')
axes[0].axis('off')

# Plot Noisy
axes[1].imshow(noisy_image, cmap='gray')
axes[1].set_title('2. Noisy')
axes[1].axis('off')

# Plot Filtered
axes[2].imshow(filtered_image, cmap='gray')
axes[2].set_title('3. Filtered')
axes[2].axis('off')

# Adjust layout and show the final masterpiece
plt.tight_layout()
plt.show()
```
