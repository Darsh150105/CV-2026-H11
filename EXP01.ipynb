```python
# Import the required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image
img = cv2.imread("image.webp")
```

```python
# Display the loaded image
cv2.imshow("Original Image", img)

# Wait until a key is pressed
cv2.waitKey(0)

# Close all OpenCV windows
cv2.destroyAllWindows()
```

```python
# Display the dimensions of the image
print("Image Shape:", img.shape)
```

```python
# Display the data type of the image
print("Image Data Type:", img.dtype)
```

```python
# Crop the image
cropped_img = img[100:400, 150:500]

# Display the cropped image
cv2.imshow("Cropped Image", cropped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Resize the image
resized_img = cv2.resize(img, (400, 300))

# Display the resized image
cv2.imshow("Resized Image", resized_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Change the color of a single pixel
img[100, 150] = [0, 0, 0]

# Display the modified image
cv2.imshow("Modified Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Convert the color image to grayscale
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
cv2.imshow("Grayscale Image", gray_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Convert the image from BGR to RGB
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the RGB image using Matplotlib
plt.imshow(rgb_img)
plt.title("RGB Image")
plt.axis("off")
plt.show()
```

```python
# Extract the red channel from the image
red_channel = img[:, :, 2]

# Display the red channel
cv2.imshow("Red Channel", red_channel)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Flip the image horizontally
flipped_img = cv2.flip(img, 1)

# Display the flipped image
cv2.imshow("Flipped Image", flipped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Rotate the image by 90 degrees clockwise
rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

# Display the rotated image
cv2.imshow("Rotated Image", rotated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Apply Gaussian Blur to the image
blurred_img = cv2.GaussianBlur(img, (7, 7), 0)

# Display the blurred image
cv2.imshow("Gaussian Blurred Image", blurred_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
