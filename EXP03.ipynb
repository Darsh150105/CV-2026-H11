```python
import cv2
import numpy as np
```

```python
img = cv2.imread("image.webp")

cv2.imshow("Original Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
def add_salt_pepper_noise(image, amount=0.03):
    noisy = image.copy()

    rows, cols = image.shape[:2]

    num_salt = int(amount * rows * cols / 2)

    # Salt
    for i in range(num_salt):
        x = np.random.randint(0, rows)
        y = np.random.randint(0, cols)
        noisy[x, y] = 255

    # Pepper
    for i in range(num_salt):
        x = np.random.randint(0, rows)
        y = np.random.randint(0, cols)
        noisy[x, y] = 0

    return noisy

noisy = add_salt_pepper_noise(img)

cv2.imshow("Noisy Image", noisy)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
def mean_filter(image, kernel_size=3, stride=1):
    
    pad = kernel_size // 2

    padded = cv2.copyMakeBorder(
        image,
        pad,
        pad,
        pad,
        pad,
        cv2.BORDER_REFLECT
    )

    output = np.zeros_like(image)

    h, w = image.shape[:2]

    for i in range(0, h, stride):
        for j in range(0, w, stride):

            roi = padded[i:i+kernel_size, j:j+kernel_size]

            mean = np.mean(roi, axis=(0,1))

            output[i, j] = mean

    return output
```

```python
def median_filter(image, kernel_size=3, stride=1):
    
    pad = kernel_size // 2

    padded = cv2.copyMakeBorder(
        image,
        pad,
        pad,
        pad,
        pad,
        cv2.BORDER_REFLECT
    )

    output = np.zeros_like(image)

    h, w = image.shape[:2]

    for i in range(0, h, stride):
        for j in range(0, w, stride):

            roi = padded[i:i+kernel_size, j:j+kernel_size]

            median = np.median(roi, axis=(0,1))

            output[i, j] = median

    return output
```

```python
mean_3_1 = mean_filter(noisy, kernel_size=3, stride=1)

cv2.imshow("Mean Filter (3x3, Stride=1)", mean_3_1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
mean_3_2 = mean_filter(noisy, kernel_size=3, stride=2)

cv2.imshow("Mean Filter (3x3, Stride=2)", mean_3_2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
mean_5_1 = mean_filter(noisy, kernel_size=5, stride=1)

cv2.imshow("Mean Filter (5x5, Stride=1)", mean_5_1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
median_3_1 = median_filter(noisy, kernel_size=3, stride=1)

cv2.imshow("Median Filter (3x3, Stride=1)", median_3_1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
median_5_1 = median_filter(noisy, kernel_size=5, stride=1)

cv2.imshow("Median Filter (5x5, Stride=1)", median_5_1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
