```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the image directly in grayscale
image_path = 'image.webp'
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Check if the image loaded correctly
if img is None:
    print("Error: Could not load the image. Check the filename!")
else:
    print("Step 2 Complete: Image loaded successfully! Dimensions:", img.shape)
    
    # Display the image to verify
    plt.figure(figsize=(6, 6))
    plt.imshow(img, cmap='gray')
    plt.title('Grayscale Image')
    plt.axis('off')
    plt.show()
```

```python
# 1. Calculate gradient in the X direction (detects vertical edges)
# We use cv2.CV_64F to prevent data loss (overflow) when gradients are negative
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobelx_abs = cv2.convertScaleAbs(sobelx) # Convert back to uint8

# 2. Calculate gradient in the Y direction (detects horizontal edges)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobely_abs = cv2.convertScaleAbs(sobely) # Convert back to uint8

# 3. Combine the X and Y gradients to get the final edge image
sobel_combined = cv2.addWeighted(sobelx_abs, 0.5, sobely_abs, 0.5, 0)

print("Step 3 Complete: Sobel edges calculated!")

# 4. Plot all three side-by-side to see the process
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(sobelx_abs, cmap='gray')
plt.title('Sobel X (Vertical Edges)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(sobely_abs, cmap='gray')
plt.title('Sobel Y (Horizontal Edges)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Combined (All Edges)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
