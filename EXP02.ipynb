```python
import cv2
import numpy as np

# Read image
img = cv2.imread("image.webp", cv2.IMREAD_GRAYSCALE)

# Function to add Salt & Pepper Noise
def add_noise(image, probability):

    noisy = image.copy()

    rows, cols = image.shape

    random_values = np.random.random((rows, cols))

    noisy[random_values < probability / 2] = 0

    noisy[random_values > (1 - probability / 2)] = 255

    return noisy

# Apply noise
noise = 0.02

noisy_image = add_noise(img, noise)

# Display result
cv2.imshow("Noisy Image", noisy_image)

cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
noise_values = [0.0, 0.33, 0.66, 1.00]

for value in noise_values:

    result = add_noise(img, value)

    cv2.imshow(f"Noise Level = {value}", result)

    cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
import cv2
import time

# Read image
img = cv2.imread("image.webp", cv2.IMREAD_GRAYSCALE)

# Create window
cv2.namedWindow("Moving Image", cv2.WINDOW_NORMAL)

# Resize window
cv2.resizeWindow("Moving Image", 300, 300)

# 1. Top Left
cv2.moveWindow("Moving Image", 0, 0)
cv2.imshow("Moving Image", img)
cv2.waitKey(1000)      # Wait 1 second

# 2. Top Right
cv2.moveWindow("Moving Image", 1000, 0)
cv2.imshow("Moving Image", img)
cv2.waitKey(1000)

# 3. Center
cv2.moveWindow("Moving Image", 500, 250)
cv2.imshow("Moving Image", img)
cv2.waitKey(1000)

# 4. Bottom Left
cv2.moveWindow("Moving Image", 0, 600)
cv2.imshow("Moving Image", img)
cv2.waitKey(1000)

# 5. Bottom Right
cv2.moveWindow("Moving Image", 1000, 600)
cv2.imshow("Moving Image", img)
cv2.waitKey(1000)

cv2.destroyAllWindows()
```
